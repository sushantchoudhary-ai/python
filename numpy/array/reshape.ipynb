## RESHAPE
* CONVERTS 1D INTO 2D AND VICE VERSA

```python
# 2D------> rows and columns
#12-------> 3 rows and 4 columns
#12--------> 4 rows and 3 columns
#AND SO ON 
np.reshape(a,(3,4)) # IT WILL RESHAPE THE ARRAY INTO 3 ROWS AND 4 COLUMNS
```

```python
b=np.random.randint(1,100,size=(2,5))
b
```

```python
np.reshape(b,10) # IT WILL RESHAPE THE ARRAY INTO 1D ARRAY WITH 10 ELEMENTS
```
