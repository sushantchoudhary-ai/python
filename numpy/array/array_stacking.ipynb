## Their are two types of stacks in array

### Horizontal stack ----------> hstack
* NO. ROWS MUST BE SAME
* IT CONTAINS DTYPE PARAMETER

```python
np.hstack((a,b))
```

```python
np.hstack((a,b),dtype=np.float64)
```

### Vertical stack --------------> vstack
* NO. OF COLUMNS MUST BE SAME

```python
np.vstack((a,b))
```

```python
np.vstack((a,b),dtype=np.float64)
```

### Stack Operator
* It can do both hstack and vstack depending on the axis we give 
* For vstack Axis=0 , hstack Axis=1 
* Process is little Bit different 

#### IN VSTACK OPERATION IT PRINTS AN ARRAY AFTER OTHER

```python
np.stack((a,b),axis=0)
```

##### IN HSTACK OPERATION HERE IT TAKES INDEX ROW 0 OF A WITH ROW 0 OF B MAKES IT A SUBARAY AND SAME WITH ALL THE ROWS 
* IT IS LITTLE BIT DIFFERENT THAN HSTACK OPERATION 

```python
np.stack((a,b),axis=1)
```
