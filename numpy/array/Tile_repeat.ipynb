```python
import numpy as np
```

```python
a=np.random.randint(1,100,size=(2,3))
print(a)    
```

## Tile 
* IF WE WANT TO REPEAT PURRA ARRAY THEN USE TILE 

```python
np.tile(a,4) # IT WILL REPEAT THE ARRAY 4 TIMES IN HORIZONTAL DIRECTION
```

```python
np.tile(a,(2,4)) # IT WILL REPEAT THE ARRAY 2 TIMES IN VERTICAL DIRECTION AND 4 TIMES IN HORIZONTAL DIRECTION
```

## Repeat
* IF WE WANT TO REPEAT AN ELEMENT N NO. OF TIMES THEN USE REPEAT

```python
np.repeat(a,4) # IT WILL REPEAT THE ELEMENTS OF ARRAY 4 TIMES IN HORIZONTAL DIRECTION
```

```python
np.repeat(a,3,axis=1) # IT WILL REPEAT THE ELEMENTS OF ARRAY 3 TIMES IN HORIZONTAL DIRECTION
```

```python
np.repeat(a,3,axis=0) # IT WILL REPEAT THE ELEMENTS OF ARRAY 3 TIMES IN VERTICAL DIRECTION
```
