```python
import numpy as np
```

```python
marks=np.random.randint(1,100,size=5)
marks
```

## WHERE ()
* IT IS USED TO ENCODE DATA AND USEED TO TRANSFORM DATA 
* IF WE HAVE TO PERFORM ONLY TWO OPERATION(IF AND ELSE VALA HISAB KITAB)

```python
np.where(marks>50) # IT WILL RETURN THE INDEX OF THE ELEMENTS WHICH ARE GREATER THAN 50
```

```python
marks[np.where(marks>50)] # IT WILL RETURN THE ELEMENTS WHICH ARE GREATER THAN 50 AS "Pass" AND LESS THAN 50 AS "Fail"
```

## SELECT
* IT IS USED WHEN WE WANT TO ASSIGN DATA BY CONSIDERING VARIOUS CONDITIONS
* IF ELIF ELSE VALA HISAB KITAB

```python
np.select([marks>=90,marks>=80,marks>=70,marks>=60,marks>=50],[ "A+","A","B","C","D"],default="F") # IT WILL RETURN THE GRADE OF THE STUDENT BASED ON THE MARKS OBTAINED
```
