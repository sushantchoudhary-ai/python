```python
import numpy as np
```

# 1 D array


## Array creation

### array()


```python
a=np.array([2,4,6,8])
print(a)
```

### arange()

```python
a=np.arange(2,10,2)
print(a)
```

### zeros()

```python
a=np.zeros(3,) #Default dtype=float to remove this do dtype=int
print(a)
```

```python
a=np.zeros(3,dtype=int)
print(a)
```

### ones()

```python
a=np.ones(3) #Default dtype=float to remove this do dtype=int
print(a)
```

```python
a=np.ones(3,dtype=int)
print(a)
```

### full()

It gives an array of required shape with a given fixed fill value 
Like if (3,5) is given it means 3 elements with 5 value

```python
a=np.full(3, 5.0) #Data type is of the filled value 
print(a)
```

### Linspace()

It gives an array of evenly spaced values over a specified interval 
Like in below example 5 equal spaced value between 1 and 10 is given 

```python
a=np.linspace(1, 10, 5) #Start, Stop, Number of points
print(a)
```

### Logspace()

Generated an array with a given number of points in a given interval but in log space 

```python
a=np.logspace(1, 10, 5) #Start, Stop, Number of points
print(a)
```

### Identity()

Gives an identity matrix of given shape

```python
a=np.identity(3,dtype=int) #Identity matrix of size 3x3
print(a) # Default dtype=float to remove this do dtype=int
```

### Eye()

This also gives an identity matrix but here we are able to move the diagonal 1 up and down with the help of a parameter 
And it can also give non square identity matrix see code third

```python
a=np.eye(3)
print(a) #Identity matrix of size 3x3
```

```python
a=np.eye(3, k=1) #Identity matrix of size 3x3 with diagonal offset by 1
print(a) # Default dtype=float to remove this do dtype=int
```

```python
a=np.eye(3,2,k=1) #Identity matrix of size 3x2 with diagonal offset by 1
print(a) # Default dtype=float to remove this do dtype=int
```

### _like family Function

They act as a cloning tool and is used to copy the shape and datatype of an existing array 

#### zeros_like()

```python
b=np.zeros_like(a) #Returns an array of zeros with the same shape and type as a
print(b)
```

#### ones_like()

```python
b=np.ones_like(a) #Returns an array of ones with the same shape and type as a
print(b)
```

#### full_like()

```python
b=np.full_like(a, 5.0) #Returns an array of filled values with the same shape and type as a
print(b)
```

### Random Array Creation

#### random.rand()

```python
a=np.random.rand(3,3) #Random numbers from uniform distribution over [0, 1)
print(a)
```

#### random.randn()

```python
a=np.random.randn(3,3) #Random numbers from standard normal distribution
print(a)
```

#### random.randint()

```python
a=np.random.randint(1, 10,size=3) 
print(a)
```

# 2D array

### Attributes of 2D array

```python
a=np.array([[1, 2, 3], [4, 5, 6]])
print(a)
```

#### shape

It returns the shape of the array in tuple form

```python
a.shape #Returns the shape of the array
#
```

#### size

```python
a.size #Returns the number of elements in the array
```

#### dtype

```python
a.dtype #Returns the data type of the array
```

#### itemsize

```python
a.itemsize #Returns the size in bytes of each element in the array
```

#### nbytes

```python
a.nbytes #Returns the total size in bytes of the array
```
