### SYNTAX TO DEFINE A SERIES 

* Identifier=pd.series(data structure,index=list of index)

```python
a=pd.Series() #EMPTY SERIES 
a
```

```python
a=pd.Series([1,2,3,4,5]) #with default index 
a
```

### TEXT HANDLING

IT IS PROCESS TO HANDLE OR MANAGE THE TEXT OR STRING DATA IN A SERIES 

```python
name=["raVi vohra","riSHu DIPke","susHANt CHOUdhary","Dhurandar RAina"]
s=pd.Series(name)
s
```

* TO HANDLE THE TEXT WE HAVE MULTIPLE METHODS RELATED TO THE TEXT OR STRING SUCH AS UPPER( ) , LOWER( ) , ISUPPER( ) ETC

```python
s.upper() #ERROR ------> WE CAN'T IMPLEMENT THESE METHODS DIRECTLY TO THE SERIES
```

--------> SO TO APPLY STRINGS METHOD , WE HAVE TO USE STR SUBMODULE WITH SERIES 

* syntax 
* seriesName.str.string_Method

```python
s.str.upper()
```

```python
s.str.title() #IT will make the text in title case means every word first word capital
```

```python
s.str.capitalize()
```

```python
words=s.str.split() # IT WILL SPLIT THE WORDS IN THE SERIES AND GIVE SERIES OF LIST OF STRING
words
```

```python
s.str.replace(" ","@") #IT REPLACE ALL THE SPACE WITH @ 
```

```python
words.str.join("@") #IT JOINS SERIES OF LIST OF STRINGS WITH THE PARAMETER GIVING INSIDE 
```

```python
s.str.join(" ") #JOIN FUNCTION WHEN USED IN STRING TAKES INDIVIDUAL ELEMENTS FROM A STRING AND GIVES A GAP BTW THEM 
```

QUESTION - CLEAN THIS RANDOM DATA GIVEN 

```python
text = pd.Series([
    "   Python@@@Programming   ",
    "\tData###Science\t",
    "**Machine   Learning**",
    "   Artificial@@@Intelligence###   ",
    "Deep---Learning!!!",
    "\nSQL___Basics\n",
    "  Pandas@@@Library###  ",
    "\tNumPy%%%Arrays\t",
    "Matplotlib***Charts!!!",
    "Data$$$Visualization###",
    "Hello@@@###World!!!",
    "Clean___This---Text",
    "**Welcome@@@to###Python**",
    "   Learn%%%Data$$$Science   ",
    "\nAI&&&ML@@@Course###\n"
])
```

```python
text.str.strip("@ \t*-_!#\n").str.replace("@"," ").str.replace("#"," ").str.replace("-"," ").str.replace("_"," ").str.replace("$"," ").str.replace("&"," ").str.replace("%"," ").str.replace("*"," ").str.replace(" "," ").str.split().str.join(" ")
```
