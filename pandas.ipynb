## SERIES

### BASIC DEFINITIONS 

* IT IS A PANDAS DATA STRUCTURE 
* IT IS UNIDIRECTIONAL 
* HAVE TWO TYPES OF INDICES 
    * DEFAULT 
    * LABELLED
* IT IS HOMOGENEOUS DATA STRUCTURE 

#### IDENTIFICARION OF SERIES


* index 1-----value 1 
* index 2-----value 2
* index 3-----value 3
* index 4-----value 4
* dtype: Datatype of values 

### SYNTAX TO DEFINE A SERIES 

* Identifier=pd.series(data structure,index=list of index)

```python
a=pd.Series() #EMPTY SERIES 
a
```

```python
a=pd.Series([1,2,3,4,5]) #with default index 
a
```

```python
a=pd.Series([1,2,3,4,5],index=['a','b','c','d','e']) # with labelled index
a
```

#### Dictionary to series

```python
marks={"Alice": 85, "Bob": 90, "Charlie": 78} #dictionary
a=pd.Series(marks) #creating series from dictionary
a
```

* IF WE WANT SERIES WITH CUSTOM INDEX WE CAN GIVE IT AND IT WILL ACCESS INDEX
* PRIORITY IS GIVEN TO INDEX 
* IF THE ELEMENT IN INDEX LIST IS NOT FOUND THEN RETURN NAN 

```python
a=pd.Series(marks,index=['Bob','Alice']) 
a
```

```python
a=pd.Series(marks,index=['Bob','Alice','David']) #David is not in the dictionary so it will be NaN
a
```
